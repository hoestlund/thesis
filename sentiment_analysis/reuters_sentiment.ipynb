### Sentiment Analysis of Reuters Data

```python
import pandas as pd
from nltk.sentiment import vader
```

```python
reuters_data = pd.read_csv('../data/reuters/combined.csv')
reuters_data = reuters_data[['title','description','date']]
reuters_data
```

```python
# Using the default vader_lexicon.txt
import nltk
nltk.download('vader_lexicon')
analyser = vader.SentimentIntensityAnalyzer()
```

```python
headline_sentiment = reuters_data['title'].apply(analyser.polarity_scores)
headline_sentiment
```

```python
reuters_data = reuters_data.join(pd.DataFrame(headline_sentiment.to_list()), how='right')
reuters_data
```

```python
description_sentiment = reuters_data['description'].apply(analyser.polarity_scores)
reuters_data = reuters_data.join(pd.DataFrame(description_sentiment.to_list()), how='right', rsuffix='_description')
reuters_data
```
